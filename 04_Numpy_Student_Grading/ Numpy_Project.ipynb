# **What You're Aiming For**

## **Objective:**

To create a program that calculates the grades of students using Numpy arrays.

# **Description:**

In this exercise, you will create a program that calculates the grades of students using Numpy arrays. The program will ask the user to enter the number of students, the number of subjects, and the marks of each student in each subject. It will then calculate the total marks, percentage, and grade for each student using Numpy arrays. The program will display the result for each student in a tabular format.


# **Instructions**
Instructions:

1. Import the Numpy library.
2. Ask the user to enter the number of students and the number of subjects.
3. Create a Numpy array to store the marks of each student in each subject. The array should have the shape (number of students, number of subjects).
4. Ask the user to enter the marks of each student in each subject.
5. Calculate the total marks for each student using the sum() function of Numpy.
6. Calculate the percentage for each student using the total marks and the formula: percentage = (total marks / (number of subjects * 100)) * 100
7. Calculate the grade for each student using the percentage and the following grading system:
1. 90% or above: A+
2. 80% or above: A
3. 70% or above: B+
4. 60% or above: B
5. 0% or above: C
6. below 50%: F

8. Display the result for each student in a tabular format with columns for student name, total marks, percentage, and grade.


```python
import numpy as np
```

```python
get_stuNo = int(input("Enter the number of students: "))

get_subNo = int(input("Enter the number of subjects: "))

box = np.zeros((get_stuNo, get_subNo))

for i in range(get_stuNo):
  for j in range(get_subNo):
    box[i,j] = int(input(f"Enter mark for Student {i+1}, Subject {j+1}: "))
total_marks = np.sum(box, axis=1)
percentage = (total_marks / (get_subNo * 100)) * 100

grades = []

for p in percentage:
  if p >= 90:
    grades.append('A+')
  elif p >= 80:
    grades.append("A")
  elif p >= 70:
    grades.append('B+')
  elif p >= 60:
    grades.append('B')
  elif p >= 50:
    grades.append('C')
  elif p < 50:
    grades.append('F')
print()
print("Student \t| Total \t| Percentage \t| Grade")
print("-" *50)
for i in range(get_stuNo):
  print(f"Student {i+1} \t| {total_marks[i]} \t| {percentage[i]:.2f}% \t| {grades[i]}")
```
